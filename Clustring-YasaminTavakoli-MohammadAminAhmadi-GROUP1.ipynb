# Outlier Detection

```python
# Outlier Detection
# IQR
def percentage_outliers(df):
    outlier_percentages = {}

    for col in df.select_dtypes(include=['float64', 'int64']).columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers_count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]

        total_count = df[col].shape[0]
        outliers_percentage = (outliers_count / total_count) * 100

        outlier_percentages[col] = outliers_percentage

    return outlier_percentages

outliers_percentage = percentage_outliers(df)
print(outliers_percentage)
```

```python
# Fix Outliers
def fix_outliers(df, outliers_percentage, threshold=5):
    cleaned_df = df.copy()

    geo_columns = ['location_latitude', 'location_longitude']

    for col, percentage in outliers_percentage.items():
        if col in geo_columns:
            continue  

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - threshold * IQR
        upper_bound = Q3 + threshold * IQR

        if percentage < threshold:
            cleaned_df = cleaned_df[~cleaned_df[col].isin(df[(df[col] < lower_bound) | (df[col] > upper_bound)].index)]

        else:
            cleaned_df[col] = cleaned_df[col].clip(lower=lower_bound, upper=upper_bound)

    return cleaned_df
```

```python
cleaned_df = fix_outliers(df, outliers_percentage)
cleaned_df
```

```python
from sklearn.decomposition import PCA, IncrementalPCA

pca_temp = PCA(n_components=0.85, svd_solver='full', random_state=42)
pca_temp.fit(X_scaled)
n_needed = pca_temp.n_components_
print("برای پوشش ۸۵٪ واریانس، روی نمونه به", n_needed, "مؤلفه نیاز داریم.")
```

```python
ipca = IncrementalPCA(n_components=n_needed)
batch_size = 100_000
n_samples = X_scaled.shape[0]

for start in range(0, n_samples, batch_size):
    end = min(start + batch_size, n_samples)
    ipca.partial_fit(X_scaled[start:end])

X_pca = np.zeros((n_samples, n_needed), dtype=np.float32)
for start in range(0, n_samples, batch_size):
    end = min(start + batch_size, n_samples)
    X_pca[start:end] = ipca.transform(X_scaled[start:end]).astype(np.float32)

print("X_pca.shape:", X_pca.shape)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pyproj import Transformer

df_small = df_final[['location_latitude', 'location_longitude', 'price_value']].copy()
df_small['cluster'] = labels.astype(int)

transformer = Transformer.from_crs("epsg:4326", "epsg:32639", always_xy=True)
utm_coords = df_small.apply(
    lambda row: transformer.transform(
        row['location_longitude'], row['location_latitude']
    ), axis=1
)
utm_coords = np.vstack(utm_coords.values)
df_small['utm_easting'] = utm_coords[:, 0]
df_small['utm_northing'] = utm_coords[:, 1]

min_size, max_size = 10, 50
price_norm = ((df_small['price_value'] - df_small['price_value'].min()) /
              (df_small['price_value'].max() - df_small['price_value'].min()))
sizes = price_norm * (max_size - min_size) + min_size

plt.figure(figsize=(12, 9))
scatter = plt.scatter(
    df_small['utm_easting'],
    df_small['utm_northing'],
    c=df_small['cluster'],
    cmap='tab10',
    s=sizes,
    alpha=0.7,
    edgecolors='k',
    linewidth=0.2
)

centers = df_small.groupby('cluster')[['utm_easting', 'utm_northing']].mean().reset_index()
plt.scatter(
    centers['utm_easting'],
    centers['utm_northing'],
    c='black',
    marker='X',
    s=200,
    label='Cluster Centers'
)

plt.title('Scatter Plot of Listings by UTM Coordinates & Price (Colored by Cluster)')
plt.xlabel('UTM Easting (meters)')
plt.ylabel('UTM Northing (meters)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []
K_range = range(1, 21)

for k in K_range:
    print(f"Fitting KMeans for k={k} ...", end=' ')
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_pca)
    inertia = kmeans.inertia_
    wcss.append(inertia)
    print(f"WCSS={inertia:.2f}")

print("\nWCSS values for k=1 to 20:")
print(wcss)

plt.figure(figsize=(8, 5))
plt.plot(K_range, wcss, marker='o')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("WCSS (Inertia)")
plt.title("Elbow Method to Determine Optimal k")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score
import numpy as np
import warnings

warnings.filterwarnings("ignore")

X_subsample = X_pca[:2000]

silhouette_scores = []
k_values = range(2, 21)

for k in k_values:
    print(f"Calculating silhouette score for k={k} ...", end=' ')
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_subsample)
    score = silhouette_score(X_subsample, labels)
    silhouette_scores.append(score)
    print(f"score = {score:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(k_values, silhouette_scores, marker='o', color='orange')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score vs k (on subsample)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd


df_final = pd.read_csv('cleaned_df.csv')
```

```python
from pyproj import Transformer
import numpy as np
df_small = df_final[['location_latitude', 'location_longitude', 'price_value']].copy()

df_small = df_small.dropna(subset=['location_latitude', 'location_longitude', 'price_value'])

transformer = Transformer.from_crs("epsg:4326", "epsg:32639", always_xy=True)
utm_coords = df_small.apply(
    lambda row: transformer.transform(row['location_longitude'], row['location_latitude']),
    axis=1
)
utm_coords = np.vstack(utm_coords.values)  
df_small['utm_easting']  = utm_coords[:, 0]
df_small['utm_northing'] = utm_coords[:, 1]
```

```python
X_raw = df_small[['utm_easting', 'utm_northing', 'price_value']].values.astype(np.float32)
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import MinMaxScaler
import numpy as np
n_total = X_raw.shape[0]   
sample_size = 100000         

rng = np.random.default_rng(42)
idx_sample = rng.choice(n_total, size=sample_size, replace=False)

X_normalized = X_raw

eps = 0.1  
min_samples = 50

db_full = DBSCAN(eps=eps, min_samples=min_samples, metric='euclidean', algorithm='kd_tree')
labels_full = db_full.fit_predict(X_normalized)
print("Number of clusters in full dataset:", len(set(labels_full)))
```

```python
n_total = X_normalized.shape[0]
n_total
```

```python
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import NearestNeighbors

core_sample_indices = db_full.core_sample_indices_
X_core = X_normalized[core_sample_indices]  
labels_core = labels_full[core_sample_indices] 

nn = NearestNeighbors(radius=eps, algorithm='ball_tree', n_jobs=1)
nn.fit(X_core)  # Fit on core points, not full dataset

batch_size = 100000
all_labels = np.full(n_total, -1, dtype=np.int32)

for start in range(0, n_total, batch_size):
    end = min(start + batch_size, n_total)
    X_batch = X_normalized[start:end]
    indices = nn.radius_neighbors(X_batch, return_distance=False)
    for i, neighbors in enumerate(indices):
        if len(neighbors) > 0:
            chosen_core = neighbors[0]  # Index into X_core
            all_labels[start + i] = labels_core[chosen_core]

print("Number of clusters in full dataset:", len(set(all_labels)))
```

```python
df_small['dbscan_cluster'] = all_labels

counts = df_small['dbscan_cluster'].value_counts(dropna=False).sort_index()
print("تعداد رکوردها در هر برچسب (DBSCAN):")
print(counts.to_dict())
```
